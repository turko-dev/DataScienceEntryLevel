## Pandas Methods
+ `pivot()`
+ `melt()`
+ `stack()`
+ `unstack()`
+ `pivot_table()`

## `pivot()`

Takes data from long to wide format and reshapes columns into rows. Takes in 3 parameters (index, columns, values)



## `melt()`

The opposite of `pivot()`. It takes data from wide to long format



## `stack()` and `unstack()`

These two functions are similar to the previous two. The difference is that they use multi-indexed data



## `pivot_table()`

This function can aggregate data whilst reshaping, and it only works on numeric data.

+ Aggregating means functions such as `sum` `count` `mean` etcetera.
