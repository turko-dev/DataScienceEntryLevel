# Data Transformations

### 1. Scaling & Normalisation

Min-Max Scaling (Normalisation)
+ Rescales data to a fixed range, usually between 0 and 1
+ Useful when the feature has a known, fixed range

Z-Score Standardisation
+ Centres the data around zero by subtracting the mean and dividing by the standard deviation
+ Useful when the feature follows a Gaussian (normal) distribution

### 2. Power Transformation (Square Root, Cube Root, Box-Cox)

+ Stabilises variance and makes data more normally distributed
+ Power transformations like square root or cube root are used to reduce skewness in data and the impact of outliers
+ Square Root
+ Cube Root
+ Box-Cox

Box-Cox Transformation
+ Stabilises variance and makes data even more evenly distributed
+ `scipy.stats.boxcox(data)`

### 3. Log Transformation

+ For data with a skewed distribution or data containing extreme values, applying a log transformation to make the data more normal
+ Reduced the influence of large outliers and compresses the range of the data - slope for logarithmic function is steeper for smaller values so expands the differences
+ Commonly used when the data includes a wide range of values

### 4. Binning (Discretisatsion)

+ Converts continuous features into categorical data
+ Useful when analysing data with lots of outliers or to categorise data into ranges

Equal-Width Binning
+ Divides data into bins of equal width

Equal-Frequency Binning
+ Divides the data such that each bin contains approximately the same number of data points
