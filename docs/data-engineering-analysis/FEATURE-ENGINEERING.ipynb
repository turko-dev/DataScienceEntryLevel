## Dimensionality Reduction (PCA)

PCA is a technique used to reduce the number of features in a dataset while retaining as much information as possible.

+ Imagine you have several interrelated features such as `temp` `humidity` and `windspeed`

+ These features might overlap in the information that they provide

+ PCA combines them into new features called principle components, which capture the most important patterns in the data

Take the example of summarising `weather_pca` from `temp`, `humidity`, and `windspeed`

## Creating Flags

You can create a binary feature such as `is_weekend` and give it a True or False value. This helps to capture patterns such as if the bike usage differs between weekdays and weekends.
